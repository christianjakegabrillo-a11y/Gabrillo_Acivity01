```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
```

```python
print(iris.feature_names)
```

```python
print(iris.target_names)
```

```python
df  = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
df["species"]= iris.target_names[iris.target]
df.head()
```

```python
plt.figure(figsize=(8,6))

for species in df["species"].unique():
    subset=df[df["species"]==species]
    plt.scatter(
        subset["sepal length (cm)"],
        subset["sepal width (cm)"],
        label=species # Added label for legend
    )
plt.xlabel("sepal length (cm)") # Corrected column name
plt.ylabel("sepal width (cm)") # Changed to ylabel and corrected column name
plt.title("iris species based on sepal measurements") # Corrected typo in 'measurements'
plt.legend()
plt.show()
```

```python
  x=iris.data
  y= iris.target
```

```python
x_train, x_test, y_train, y_test=train_test_split(
x,
y,
test_size=0.20,
random_state=42,
stratify=y
)
print("training obeservation:",len(x_train))
print("testing obeservation:",len(x_test ))
```

```python
model= KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(x_train, y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
print(y_pred)
```

```python
print("Actual: ")
print(y_test)

print("\nPredicted: ")
print(y_pred)
```

```python
accuracy=accuracy_score(y_test, y_pred)
print("accuracy: ", accuracy)
```

```python
new_flower = [[5.1, 3.5, 1.4, 0.2]]
```

```python
prediction=model.predict(new_flower)
```

```python
print(
    "predicted species: ",
    iris.target_names[prediction[0]]
)
```

```python
probabilities =model.predict_proba(new_flower)
print(probabilities)
```

```python
for species, probability in zip(
    iris.target_names,
    probabilities[0]
):
    print(f"{species}: {probability:.2%}") # Corrected indentation and f-string format
```

```python
k_values = [1,3,5,7,9]
accuracies =[]
for k in k_values:
    model=KNeighborsClassifier(n_neighbors=k) # Added colon, corrected kwargs
    model.fit(x_train, y_train)
    y_pred =model.predict(x_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracies.append(accuracy)
    print(f"k={k}, accuracy={accuracy:.2f}") # Corrected indentation
```

```python
plt.figure(figsize = (8,5))
plt.plot(
    k_values,
    accuracies,
    marker="o" # Changed '0' to 'o'
)
plt.xlabel("k value")
plt.ylabel("Accuracy") # Changed to ylabel and corrected label
plt.title("KNN model accuracy for different k values") # Corrected typo

plt.xticks(k_values)
plt.show()
```
